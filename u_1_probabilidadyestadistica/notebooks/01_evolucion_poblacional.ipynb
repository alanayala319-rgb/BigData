## Actividad 3 — Estadística descriptiva

**Propósito:** Aplicar medidas estadísticas básicas a una variable poblacional.

### Instrucciones

Selecciona un año del periodo histórico y filtra los datos correspondientes. Sobre la variable de **cambio anual de población**, calcula: media, mediana, varianza, desviación estándar, mínimo y máximo.

```python
anio = 2023  # elige un año del periodo histórico
columna_poblacion = "Annual change in population"

df_anio = df[df["Year"] == anio]
df_anio.head()
```

```python
print(df.columns.tolist())
```

```python
media = df_anio[columna_poblacion].mean()
mediana = df_anio[columna_poblacion].median()
varianza = df_anio[columna_poblacion].var()
desviacion_estandar = df_anio[columna_poblacion].std()
minimo = df_anio[columna_poblacion].min()
maximo = df_anio[columna_poblacion].max()

print("Media:", media)
print("Mediana:", mediana)
print("Varianza:", varianza)
print("Desviación estándar:", desviacion_estandar)
print("Mínimo:", minimo)
print("Máximo:", maximo)
```

```python
# Entidades con el valor mínimo y máximo
entidad_min = df_anio.loc[df_anio[columna_poblacion].idxmin(), "Entity"]
entidad_max = df_anio.loc[df_anio[columna_poblacion].idxmax(), "Entity"]

print("Entidad con valor mínimo:", entidad_min)
print("Entidad con valor máximo:", entidad_max)
```
